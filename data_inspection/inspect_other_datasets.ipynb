```python
merged_df['ts']        = pd.to_datetime(merged_df['ts'])
merged_df['time']      = merged_df['ts'].astype('int64') // 10**9         # epoch secs -> time_col
```

```python
# numeric node features
merged_df['log_paid']  = np.log1p(merged_df['amt_paid'])
merged_df['log_recv']  = np.log1p(merged_df['amt_recv'])
merged_df['amt_ratio'] = (merged_df['amt_recv'] / merged_df['amt_paid'].replace(0,np.nan)).fillna(1)
merged_df['hour']      = merged_df['ts'].dt.hour
```

```python
for c in ['cur_paid','cur_recv','fmt']:                      # frequency-encode
    merged_df[c+'_FE'] = merged_df[c].map(merged_df[c].value_counts())
```

```python
merged_df
```

```python
bank_name_map = (
    accounts.drop_duplicates("bank_id")
    .set_index("bank_id")["bank_name"]
)

transactions = merged_df.assign(
    from_bank_name=merged_df["from_bank"].map(bank_name_map),
    to_bank_name=merged_df["to_bank"].map(bank_name_map),
)
```

```python
mask = (
    transactions.groupby("from_acct")["from_bank"]
    .transform("nunique")
    .gt(1)
)

different_source_bank_pairs = (
    transactions.loc[
        mask,
        ["from_acct", "from_bank", "from_bank_name"]
    ]
    .drop_duplicates()
    .sort_values(["from_acct", "from_bank"])
)

different_source_bank_pairs
```

```python
different_source_bank_summary = (
    different_source_bank_pairs.groupby("from_acct")
    .agg(
        bank_count=("from_bank", "size"),
        bank_ids=("from_bank", list),
        bank_names=("from_bank_name", list),
    )
    .reset_index()
)
different_source_bank_summary
```

```python
mask = (
    transactions.groupby("to_acct")["to_bank"]
    .transform("nunique")
    .gt(1)
)

different_to_bank_pairs = (
    transactions.loc[
        mask,
        ["to_acct", "to_bank", "to_bank_name"]
    ]
    .drop_duplicates()
    .sort_values(["to_acct", "to_bank"])
)

different_to_bank_pairs
```

```python
different_to_bank_summary = (
    different_to_bank_pairs.groupby("to_acct")
    .agg(
        bank_count=("to_bank", "size"),
        bank_ids=("to_bank", list),
        bank_names=("to_bank_name", list),
    )
    .reset_index()
)
different_to_bank_summary
```

```python
equal_rows_same_currency = merged_df.loc[equal_mask & same_currency].copy()

smaller_rows_same_currency = merged_df.loc[
    (merged_df["amt_recv"] < merged_df["amt_paid"]) & ~equal_mask & same_currency
].copy()

larger_rows_same_currency = merged_df.loc[
    (merged_df["amt_recv"] > merged_df["amt_paid"]) & ~equal_mask & same_currency
].copy()
```

```python
pd.Series({
    "equal": len(equal_rows_same_currency),
    "amt_recv < amt_paid": len(smaller_rows_same_currency),
    "amt_recv > amt_paid": len(larger_rows_same_currency),
})
```

```python
import pandas as pd, numpy as np
df = pd.read_csv("/Volumes/SandiskSSD/Developer/AI_Document/Financial_Fraud_Detection_Thesis/data/card_transaction_v1.csv")
```

```python
df
```

```python
df.User.value_counts()
```

```python
user_less_than_14_transactions = df[
    df.groupby("User")["User"].transform("size") < 14
]
```

```python
user_less_than_14_transactions.shape
```

```python
USER_COL = "User"

tmp = df.reset_index(names="original_index").copy()

tmp["run_id"] = tmp[USER_COL].ne(tmp[USER_COL].shift()).cumsum()

continuous_user_runs = (
    tmp.groupby(["run_id", USER_COL], dropna=False)
    .agg(
        start_index=("original_index", "first"),
        end_index=("original_index", "last"),
        continuous_rows=(USER_COL, "size")
    )
    .reset_index()
)

continuous_user_runs
```

```python
user_continuity_summary = (
    continuous_user_runs.groupby(USER_COL, dropna=False)
    .agg(
        num_blocks=("continuous_rows", "size"),
        max_continuous_rows=("continuous_rows", "max"),
        total_rows=("continuous_rows", "sum"),
        avg_continuous_rows=("continuous_rows", "mean")
    )
    .sort_values("max_continuous_rows", ascending=False)
)

user_continuity_summary
```

```python
import pandas as pd, numpy as np
sparkov_df = pd.read_csv("/Volumes/SandiskSSD/Developer/AI_Document/Financial_Fraud_Detection_Thesis/data/Simulated_Credit_Card_Sparkov/fraudTrain.csv")
```

```python
sparkov_df
```

```python
sparkov_df["cc_num"].value_counts()
```

```python
USER_COL = "cc_num"

tmp = (
    sparkov_df
    .reset_index(names="original_index")
    .sort_values(["unix_time", "original_index"], kind="stable")
    .reset_index(drop=True)
)

tmp["run_id"] = tmp[USER_COL].ne(tmp[USER_COL].shift()).cumsum()

continuous_user_runs = (
    tmp.groupby("run_id", sort=False)
    .agg(
        cc_num=(USER_COL, "first"),
        start_index=("original_index", "first"),
        end_index=("original_index", "last"),
        start_time=("trans_date_trans_time", "first"),
        end_time=("trans_date_trans_time", "last"),
        consecutive_rows=(USER_COL, "size"),
    )
    .reset_index(drop=True)
)
```
