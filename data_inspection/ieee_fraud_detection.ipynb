# 1st Place Solution - Part 1

Time is not most important
After Microsoft Malware, I was reluctant to enter another time series competition. To my relief, this competition actually isn't about time. The reason adversarial validation has AUC=1 is not because the nature of fraud changes radically over time but rather because the clients in the dataset change radically over time.

Once you realize this, you will know that the challenge in this competition is building a model that can predict unseen clients (not unseen time). Below is a plot showing the data in the private dataset (which has 400,000 rows). 68.2% of the data are clients we do not see in train dataset. 16.4% of data are in both the train and test dataset. And 15.4% we were unsure about. Each blue line below is one client. Each client makes many transactions, the left end of the blue line is their first transaction in time and the right end is the last transaction in time.

```
# Count the number of unique clients (cards)
unique_combinations = train_transaction[['card1', 'card2', 'card3', 'card4', 'card5', 'card6']].value_counts()
print(unique_combinations)
print(len(unique_combinations))
uids_2plus = unique_combinations[unique_combinations >= 2].index
print(uids_2plus)

```

```
# Number of unique cards (clients)

cols = ['card1','card2','card3','card4','card5','card6']
n_distinct = train_transaction[cols].value_counts(dropna=False).size
print(n_distinct)


# Number of cards that appear in 2 or more transactions
counts = train_transaction[cols].value_counts(dropna=False)

n_clients_2plus = (counts >= 2).sum()
n_rows_from_those_clients = counts[counts >= 2].sum()
print(n_rows_from_those_clients)
```

```
train = train_transaction.copy()

# common preparation seen in the discussion/notebook family
train['day'] = train['TransactionDT'] / (24 * 60 * 60)
train['card1_addr1'] = train['card1'].astype(str) + '_' + train['addr1'].astype(str)
# train['uid'] = train['card1_addr1'] + '_' + np.floor(train['day'] - train['D1']).astype(str)
train['uid'] = train['card1'].astype(str) + '_' + train['card2'].astype(str) + '_' + train['card3'].astype(str) + '_' + train['card4'] + '_' + train['card5'].astype(str) + '_' + train['card6']

vc = train['uid'].value_counts()

uids_2plus = vc[vc >= 2].index
tmp = train[train['uid'].isin(uids_2plus)].groupby('uid')['isFraud'].agg(['count', 'sum', 'nunique'])

n_clients_2plus = len(tmp)  # -> 73838
n_rows_in_these_clients = tmp['count'].sum()  # -> 280829

nonfraud_uid = ((tmp['nunique'] == 1) & (tmp['sum'] == 0)).sum()              # -> 71575
fraud_uid    = ((tmp['nunique'] == 1) & (tmp['sum'] == tmp['count'])).sum()   # -> 2134
mixed_uid    = (tmp['nunique'] == 2).sum()                                     # -> 129
```

## EDA

For the first 150 columns, we used Alijs's great EDA

```
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

x_train = train_transaction["TransactionDT"]
x_test  = test_transaction["TransactionDT"]

# Use same bins for fair comparison
bins = np.linspace(min(x_train.min(), x_test.min()),
                   max(x_train.max(), x_test.max()),
                   120)

plt.figure(figsize=(12, 5))
sns.histplot(x_train, bins=bins, color="royalblue", alpha=0.45, label="Train", stat="count")
sns.histplot(x_test,  bins=bins, color="tomato",    alpha=0.45, label="Test",  stat="count")

plt.xlabel("TransactionDT")
plt.ylabel("Frequency")
plt.title("Train vs Test TransactionDT Distribution")
plt.legend()
plt.tight_layout()
plt.show()
```

```
corr_table = (
    train_transaction.corr(numeric_only=True)["TransactionDT"]
    .drop("TransactionDT")
    .sort_values(key=lambda s: s.abs(), ascending=False)
    .rename("Correlation with TransactionDT")
    .reset_index()
    .rename(columns={"index": "Column"})
)

corr_table.head(20)
```

```
print(train_transaction['isFraud'].value_counts())

no_of_legit_transactions = (train_transaction['isFraud'] == 0).sum()
no_of_fraud_transactions = (train_transaction['isFraud'] == 1).sum()

print("Legit transactions: ", no_of_legit_transactions)
print("Fraud transactions: ", no_of_fraud_transactions)
print("The percentage of legit transactions: ", no_of_legit_transactions / len(train_transaction))
print("The percentage of fraud transactions: ", no_of_fraud_transactions / len(train_transaction))
```

```
corr_table = (
    train_transaction.corr(numeric_only=True)["isFraud"]
    .drop("isFraud")
    .sort_values(key=lambda s: s.abs(), ascending=False)
    .rename("Correlation with isFraud")
    .reset_index()
    .rename(columns={"index": "Column"})
)

corr_table.head(20)
```

```
corr_table = (
    train_transaction.corr(numeric_only=True)["TransactionID"]
    .drop("TransactionID")
    .sort_values(key=lambda s: s.abs(), ascending=False)
    .rename("Correlation with TransactionID")
    .reset_index()
    .rename(columns={"index": "Column"})
)

corr_table.head(20)
```

```
train_plot = train_transaction[["TransactionDT", "TransactionAmt", "isFraud"]].dropna()
test_plot = test_transaction[["TransactionDT", "TransactionAmt"]].dropna()

# Optional sampling for speed
train_plot = train_plot.sample(min(len(train_plot), 80000), random_state=42)
test_plot = test_plot.sample(min(len(test_plot), 80000), random_state=42)

plt.figure(figsize=(11, 6))

# Train (colored by isFraud)
sns.scatterplot(
    data=train_plot,
    x="TransactionDT",
    y="TransactionAmt",
    hue="isFraud",
    palette={0: "steelblue", 1: "crimson"},
    alpha=0.35,
    s=12
)

# Test (all green, unlabeled fraud status)
sns.scatterplot(
    data=test_plot,
    x="TransactionDT",
    y="TransactionAmt",
    color="green",
    alpha=0.25,
    s=12,
    label="Test (unlabeled)"
)

plt.xlabel("TransactionDT")
plt.ylabel("TransactionAmt")
plt.title("TransactionDT vs TransactionAmt (Train + Test)")
plt.legend(title="Class")
plt.tight_layout()
plt.show()
```
