# REFACTORED EXPERIMENT PIPELINE — IBM AML (HI-Small)
# Laundering detection: **Graph vs XGBoost**, in **two graph formulations**
# ════════════════════════════════════════════════════════════════════

Everything **below this divider is a self-contained pipeline**. It re-reads
`HI-Small_Trans.csv` and does its own feature engineering, so it does **not**
depend on any of the exploratory cells above — you can `Run All from here`.

## What gets run

The label `is_laundering` lives on the **transaction**. Every experiment below
predicts that *same* transaction-level label, on the *same rows* and the *same*
temporal train / val / test split, so all numbers are directly comparable.

| # | Experiment | Graph nodes | Runs |
|---|------------|-------------|------|
| 1 | **XGBoost alone** (baseline) | — | once |
| 2 | **Approach A — account graph** (*the appropriate AML graph*) | accounts | graph-alone · graph+XGB |
| 3 | **Approach B — transaction graph** (*your GTAN architecture*) | transactions | graph-alone · graph+XGB |

* **graph alone** = the GNN's own classifier head.
* **graph + XGBoost** = freeze the GNN, feed its node embeddings (+ raw features) to XGBoost.

## Why two graph formulations?

* **Approach A (appropriate for AML).** An AML transaction is literally a
  *directed edge* `from_account → to_account`. The graph that respects the data
  is therefore **nodes = accounts, edges = transactions**, and laundering
  detection becomes **edge classification**. Message passing learns one
  embedding per account from its counterparty neighbourhood; each transaction is
  then scored from its two endpoint embeddings plus its own edge features.

* **Approach B (your GTAN graph).** Your IEEE-Fraud GTAN treats **each
  transaction as a node** and links transactions that share an *identity key*
  (there: `uid`, `card1_addr1`, …). The AML analogue of an identity key is the
  **shared account**: two transactions are linked when they share the same
  sender (`from_id`) or receiver (`to_id`), chained in time exactly like
  `_chain_pairs` in your notebooks. We reuse your `GTANEdge` label-aware
  `TransformerConv`, the 11-dim edge attributes (recency decay / relation flags /
  relation count / cosine), top-k pruning, and weighted neighbour sampling.

## Label-reveal protocol (Approach B)

Your two attached GTAN notebooks differ only in *which node labels are revealed*
to message passing. We default to the **reference-node** protocol (the more
conservative one you leaned toward): in each fold only a fixed fraction of
*training* labels are revealed as "reference" nodes; query-train / val / test
labels stay masked, and a **reference-safe extraction** guarantees a row's own
label is never baked into its own embedding. Set `LABEL_PROTOCOL = "full"` to
instead reveal *all* historical training labels (your other notebook).

## Scale note (important — please read)

HI-Small has **~5.08M transactions** and only **~5.2k laundering (0.10 %)**.
A kNN similarity graph over 5M transaction-nodes is `O(N²)` and infeasible, so
the pipeline **sub-samples**: it keeps **all** laundering rows and down-samples
negatives to `TARGET_N` rows (default **200k**). All three experiments share the
same sample and split, which keeps the comparison fair. Flip
`USE_FULL_DATA = True` to disable sub-sampling — recommended only for Approach A
on a GPU; Approach B's kNN edges auto-disable above `SIM_MAX_NODES` nodes.

## 2 · Load + feature engineering (self-contained)

* Re-reads the raw CSV and rebuilds globally-unique account ids
  `from_id = "{bank}_{account}"` (same trick you used in the EDA).
* **Sub-samples** keeping every positive.
* **Temporal split** by timestamp quantiles (train is the earliest 70 %).
* All encodings (currency/format frequency, account degree) are fit on the
  **train split only** → no information leaks from val/test.

Outputs reused by every experiment:
`X_raw` (for XGBoost), `X_std` (standardized, for the GNNs), `y`, the split
masks, `time_arr`, and the account-id columns.

```python
# ===== Load + feature engineering =====
RAW = pd.read_csv(DATA_CSV)
RAW.columns = ['ts','from_bank','from_acct','to_bank','to_acct','amt_recv','cur_recv',
               'amt_paid','cur_paid','fmt','is_laundering']
RAW['ts']   = pd.to_datetime(RAW['ts'])
RAW['time'] = RAW['ts'].astype('int64') // 10**9
RAW['from_id'] = RAW['from_bank'].astype(str) + '_' + RAW['from_acct'].astype(str)
RAW['to_id']   = RAW['to_bank'].astype(str)   + '_' + RAW['to_acct'].astype(str)
print(f"loaded {len(RAW):,} rows | positives = {int(RAW.is_laundering.sum()):,} "
      f"({RAW.is_laundering.mean()*100:.3f}%)")

# ---- sub-sample: keep ALL positives, down-sample negatives ----
if USE_FULL_DATA:
    tx = RAW.sort_values('time').reset_index(drop=True)
else:
    pos = RAW[RAW.is_laundering == 1]
    n_neg = max(0, TARGET_N - len(pos))
    neg = RAW[RAW.is_laundering == 0].sample(n=min(n_neg, int((RAW.is_laundering == 0).sum())),
                                             random_state=SEED)
    tx = pd.concat([pos, neg]).sort_values('time').reset_index(drop=True)
print(f"working set: {len(tx):,} rows | positives = {int(tx.is_laundering.sum()):,} "
      f"({tx.is_laundering.mean()*100:.3f}%)")

# ---- temporal split by time quantiles ----
q_lo, q_hi = tx['time'].quantile(SPLIT_Q[0]), tx['time'].quantile(SPLIT_Q[1])
split   = np.where(tx['time'] <= q_lo, 'train', np.where(tx['time'] <= q_hi, 'val', 'test'))
tr_mask = split == 'train'; va_mask = split == 'val'; te_mask = split == 'test'
y = tx['is_laundering'].to_numpy().astype(np.int64)
for nm, m in [('train', tr_mask), ('val', va_mask), ('test', te_mask)]:
    print(f"  {nm:5s}: {m.sum():>8,} rows | pos = {int(y[m].sum()):>5,} ({y[m].mean()*100:.3f}%)")

# ---- per-transaction features (train-only encodings -> no leakage) ----
tx['log_paid'] = np.log1p(tx['amt_paid']); tx['log_recv'] = np.log1p(tx['amt_recv'])
tx['amt_ratio'] = (tx['amt_recv'] / tx['amt_paid'].replace(0, np.nan)).fillna(1.0)
tx['hour'] = tx['ts'].dt.hour; tx['dow'] = tx['ts'].dt.dayofweek
tx['same_cur']   = (tx['cur_recv'] == tx['cur_paid']).astype(np.int8)
tx['cross_bank'] = (tx['from_bank'] != tx['to_bank']).astype(np.int8)
tx['self_loop']  = (tx['from_id'] == tx['to_id']).astype(np.int8)

def train_freq_encode(col):
    vc = tx.loc[tr_mask, col].value_counts()
    return tx[col].map(vc).fillna(0).astype(np.float32)
for c in ['cur_paid', 'cur_recv', 'fmt']:
    tx[c + '_FE'] = train_freq_encode(c)

out_deg = tx.loc[tr_mask, 'from_id'].value_counts()
in_deg  = tx.loc[tr_mask, 'to_id'].value_counts()
tx['from_out_deg'] = np.log1p(tx['from_id'].map(out_deg).fillna(0)).astype(np.float32)
tx['to_in_deg']    = np.log1p(tx['to_id'].map(in_deg).fillna(0)).astype(np.float32)

FEAT_COLS = ['log_paid','log_recv','amt_ratio','hour','dow','same_cur','cross_bank',
             'self_loop','cur_paid_FE','cur_recv_FE','fmt_FE','from_out_deg','to_in_deg']
X_raw = tx[FEAT_COLS].to_numpy().astype(np.float32)              # for XGBoost (trees: no scaling needed)
mu = X_raw[tr_mask].mean(0); sd = X_raw[tr_mask].std(0) + 1e-6
X_std = ((X_raw - mu) / sd).astype(np.float32)                   # standardized -> GNN node features
time_arr = tx['time'].to_numpy().astype(np.float64)
NF = X_raw.shape[1]
print("feature matrix:", X_raw.shape, "| cols:", FEAT_COLS)
```

```python
# ===== small helpers =====
def report(name, y_true, prob):
    auc = roc_auc_score(y_true, prob); ap = average_precision_score(y_true, prob)
    RESULTS[name] = dict(test_auc=float(auc), test_ap=float(ap))
    print(f"  >>> [{name}]  test ROC-AUC = {auc:.4f} | test PR-AUC (AP) = {ap:.4f}")
    return auc, ap
```

## 3 · Experiment 1 — XGBoost alone (baseline)

Plain gradient-boosted trees on the per-transaction tabular features. Class
imbalance is handled with `scale_pos_weight`; we early-stop on the val split and
report on test. This is the reference every graph model must beat.

```python
# ===== Experiment 1 — XGBoost alone =====
def run_xgb(Xtr, ytr, Xva, yva, Xte, yte, tag):
    spw = float((ytr == 0).sum()) / max(1.0, float((ytr == 1).sum()))
    dtr = xgb.DMatrix(Xtr, label=ytr); dva = xgb.DMatrix(Xva, label=yva); dte = xgb.DMatrix(Xte, label=yte)
    params = {**XGB_PARAMS, 'objective': 'binary:logistic', 'scale_pos_weight': spw, 'seed': SEED}
    bst = xgb.train(params, dtr, num_boost_round=XGB_ROUNDS,
                    evals=[(dtr, 'train'), (dva, 'val')],
                    early_stopping_rounds=XGB_EARLY, verbose_eval=False)
    best = getattr(bst, "best_iteration", XGB_ROUNDS - 1)
    prob = bst.predict(dte, iteration_range=(0, best + 1))
    print(f"  [{tag}] best_iter = {best}")
    return prob, bst

prob, _ = run_xgb(X_raw[tr_mask], y[tr_mask], X_raw[va_mask], y[va_mask],
                  X_raw[te_mask], y[te_mask], "XGBoost-alone")
report("1 · XGBoost alone", y[te_mask], prob)
```

## 4 · Approach A — Account-level graph (the appropriate AML graph)

**Nodes = accounts**, **edges = transactions** (`from_id → to_id`).
The message-passing graph is built from **train-period edges only**, so
embeddings for accounts never see a future (val/test) transaction → no temporal
leakage. Each account gets train-only aggregate features (in/out counts &
mean log-amounts). We add reverse edges (with a direction flag) so receivers
also aggregate from senders.

Model: 2 × `TransformerConv` (edge-attribute aware) → account embeddings `H`.
Then an **edge classifier** scores each transaction from
`[H[from], H[to], transaction-features] → logit`.

```python
# ===== Approach A — build the account graph =====
accounts = pd.Index(pd.unique(pd.concat([tx['from_id'], tx['to_id']], ignore_index=True)))
acc2i = {a: i for i, a in enumerate(accounts)}
n_acc = len(accounts)
src_all = tx['from_id'].map(acc2i).to_numpy().astype(np.int64)
dst_all = tx['to_id'].map(acc2i).to_numpy().astype(np.int64)
print(f"account nodes: {n_acc:,} | transaction edges: {len(tx):,}")

# ---- account node features from TRAIN edges only ----
sub = tx[tr_mask]
out_cnt = sub.groupby('from_id').size(); in_cnt = sub.groupby('to_id').size()
out_amt = sub.groupby('from_id')['log_paid'].mean(); in_amt = sub.groupby('to_id')['log_recv'].mean()
node_feat = np.zeros((n_acc, 4), np.float32)
for s, col, tf in [(out_cnt, 0, True), (in_cnt, 1, True), (out_amt, 2, False), (in_amt, 3, False)]:
    idx = np.array([acc2i[a] for a in s.index], dtype=np.int64)
    node_feat[idx, col] = np.log1p(s.to_numpy()) if tf else s.to_numpy()
node_feat = ((node_feat - node_feat.mean(0)) / (node_feat.std(0) + 1e-6)).astype(np.float32)

# ---- message-passing graph = TRAIN edges, bidirectional (+ direction flag on edge_attr) ----
tr_e = np.where(tr_mask)[0]
ei_fwd = np.stack([src_all[tr_e], dst_all[tr_e]])
ei_rev = np.stack([dst_all[tr_e], src_all[tr_e]])
ea_fwd = np.concatenate([X_std[tr_e], np.ones((len(tr_e), 1), np.float32)], 1)
ea_rev = np.concatenate([X_std[tr_e], np.zeros((len(tr_e), 1), np.float32)], 1)
mp_edge_index = torch.tensor(np.concatenate([ei_fwd, ei_rev], 1), dtype=torch.long)
mp_edge_attr  = torch.tensor(np.concatenate([ea_fwd, ea_rev], 0), dtype=torch.float32)
print(f"message-passing edges (incl. reverse): {mp_edge_index.shape[1]:,} | "
      f"node_feat {node_feat.shape} | edge_attr dim {mp_edge_attr.shape[1]}")
```

```python
# ===== Approach A — model =====
class AccountGNN(nn.Module):
    def __init__(self, in_node, mp_edge_dim, txn_feat_dim,
                 hidden=A_HIDDEN, heads=A_HEADS, layers=A_LAYERS, drop=A_DROP):
        super().__init__()
        self.convs = nn.ModuleList(); self.norms = nn.ModuleList()
        dim = in_node
        for _ in range(layers):
            self.convs.append(TransformerConv(dim, hidden, heads=heads, concat=True,
                                              beta=True, dropout=drop, edge_dim=mp_edge_dim))
            self.norms.append(nn.LayerNorm(hidden * heads))
            dim = hidden * heads
        self.act = nn.PReLU(); self.drop = nn.Dropout(drop)
        self.emb_dim = dim
        self.edge_head = nn.Sequential(
            nn.Linear(2 * dim + txn_feat_dim, dim), nn.BatchNorm1d(dim), nn.PReLU(),
            nn.Dropout(drop), nn.Linear(dim, 1),
        )
    def encode(self, x, edge_index, edge_attr):
        h = x
        for conv, norm in zip(self.convs, self.norms):
            h = self.drop(self.act(norm(conv(h, edge_index, edge_attr=edge_attr))))
        return h
    def classify(self, H, src, dst, txn_feat):
        return self.edge_head(torch.cat([H[src], H[dst], txn_feat], dim=1)).squeeze(-1)

print("AccountGNN defined")
```

```python
# ===== Approach A — train (graph alone) =====
def train_account_gnn():
    torch.manual_seed(SEED)
    model = AccountGNN(node_feat.shape[1], mp_edge_attr.shape[1], NF).to(device)
    x  = torch.tensor(node_feat, device=device)
    ei = mp_edge_index.to(device); ea = mp_edge_attr.to(device)
    src_t = torch.tensor(src_all, device=device); dst_t = torch.tensor(dst_all, device=device)
    Xe = torch.tensor(X_std, device=device); yt = torch.tensor(y, device=device, dtype=torch.float32)
    tr_i = torch.tensor(np.where(tr_mask)[0], device=device)
    va_i = torch.tensor(np.where(va_mask)[0], device=device)
    te_i = torch.tensor(np.where(te_mask)[0], device=device)

    pos = float(y[tr_mask].sum()); neg = float((~(y[tr_mask] == 1)).sum())
    pw = torch.tensor([np.sqrt(neg / max(pos, 1.0))], device=device)
    loss_fn = nn.BCEWithLogitsLoss(pos_weight=pw)
    opt = torch.optim.AdamW(model.parameters(), lr=A_LR, weight_decay=A_WD)

    best_auc, best_state, bad = -1.0, None, 0
    for ep in range(1, A_EPOCHS + 1):
        model.train(); opt.zero_grad()
        H = model.encode(x, ei, ea)
        loss = loss_fn(model.classify(H, src_t[tr_i], dst_t[tr_i], Xe[tr_i]), yt[tr_i])
        loss.backward(); nn.utils.clip_grad_norm_(model.parameters(), 2.0); opt.step()
        model.eval()
        with torch.no_grad():
            H = model.encode(x, ei, ea)
            vp = torch.sigmoid(model.classify(H, src_t[va_i], dst_t[va_i], Xe[va_i])).cpu().numpy()
        va_auc = roc_auc_score(y[va_mask], vp)
        print(f"   ep {ep:2d}/{A_EPOCHS}  loss={loss.item():.4f}  val_auc={va_auc:.4f}")
        if va_auc > best_auc:
            best_auc, best_state, bad = va_auc, copy.deepcopy(model.state_dict()), 0
        else:
            bad += 1
            if bad >= A_PATIENCE:
                print(f"   early stop at epoch {ep}"); break
    model.load_state_dict(best_state)
    model.eval()
    with torch.no_grad():
        H = model.encode(x, ei, ea)
        tp = torch.sigmoid(model.classify(H, src_t[te_i], dst_t[te_i], Xe[te_i])).cpu().numpy()
        H_np = H.detach().cpu().numpy()
    return tp, H_np

A_test_prob, A_H = train_account_gnn()
report("2a · Account-graph alone", y[te_mask], A_test_prob)
```

## 5 · Approach B — Transaction-level graph (YOUR GTAN architecture)

This is a faithful port of your reference-node GTAN notebook, adapted to AML:

* **Nodes = transactions**, features = the standardized tabular matrix `X_std`.
* **Identity-chain edges**: transactions sharing the same `from_id` (relation 0)
  or `to_id` (relation 1) are chained in time with `_chain_pairs`
  (`EDGE_PER_TRANS` steps). These replace the `uid` / `card1_addr1` chains.
* **Optional kNN cosine edges** (relation 4) — auto-off above `SIM_MAX_NODES`.
* **11-dim `edge_attr`**: 3 recency-decay channels, 5 relation flags, a
  self-loop flag, a relation count, and cosine — identical layout to your
  notebook. Top-k pruning per destination + weighted neighbour sampling.
* **`GTANEdge`** = your label-aware `TransformerConv` stack.
* **Reference-node label protocol** with reference-safe embedding extraction
  (set `LABEL_PROTOCOL="full"` to reveal all train labels instead).

The cell below defines the helpers (verbatim structure from your notebook); the
next cell builds the graph and trains.

```python
# ===== Approach B — GTAN helpers (ported from your reference-node notebook) =====
REL_INDEX = {"from_id": 0, "to_id": 1, "similarity": 4}   # 5 relation-flag slots; AML uses 0,1,4

def _valid_id(v):
    v = np.asarray(v); ok = v != -1
    if np.issubdtype(v.dtype, np.floating): ok &= ~np.isnan(v)
    return ok

def _chain_pairs(labels, timestamps, edge_per_trans, valid=None):
    positions = np.arange(len(labels), dtype=np.int64)
    groups = np.asarray(labels); times_local = np.asarray(timestamps)
    if valid is not None:
        positions, groups, times_local = positions[valid], groups[valid], times_local[valid]
    if len(positions) == 0:
        empty = np.array([], dtype=np.int64); return empty, empty
    order = np.argsort(times_local, kind="mergesort")
    frame = pd.DataFrame({"node": positions[order], "group": groups[order]})
    src_list, dst_list = [], []
    for _, gframe in frame.groupby("group", sort=False):
        idx = gframe["node"].to_numpy(); glen = len(idx)
        for step in range(1, edge_per_trans):
            if step >= glen: break
            src_list.append(idx[:glen - step]); dst_list.append(idx[step:])
    if src_list: return np.concatenate(src_list), np.concatenate(dst_list)
    empty = np.array([], dtype=np.int64); return empty, empty

@torch.no_grad()
def knn_sim(X_norm_np, timestamps, uid_codes_np, k, dev, batch=512):
    n = X_norm_np.shape[0]
    Xn = torch.from_numpy(X_norm_np).to(dev)
    tt = torch.tensor(timestamps, device=dev)
    uc = torch.tensor(uid_codes_np, device=dev)
    src_parts, dst_parts = [], []
    for start in range(0, n, batch):
        end = min(start + batch, n)
        sims = Xn[start:end] @ Xn.T
        qt = tt[start:end].unsqueeze(1); qu = uc[start:end].unsqueeze(1)
        valid = (tt.unsqueeze(0) < qt) & (uc.unsqueeze(0) != qu)
        sims = sims.masked_fill(~valid, float("-inf"))
        kk = min(k, n)
        tv, ti = torch.topk(sims, kk, dim=1)
        mask = torch.isfinite(tv)
        rows = torch.arange(end - start, device=dev).unsqueeze(1).expand(-1, kk)[mask] + start
        cols = ti[mask]
        dst_parts.append(rows.cpu()); src_parts.append(cols.cpu())
        del sims, valid, tv, ti
    if dev.type == "cuda": torch.cuda.empty_cache()
    if not src_parts:
        empty = np.array([], dtype=np.int64); return empty, empty
    return torch.cat(src_parts).numpy(), torch.cat(dst_parts).numpy()

def prune_topk_per_destination(edge_index, edge_attr, k):
    dst = edge_index[1].cpu().numpy()
    cosine = edge_attr[:, 10].cpu().numpy()
    strength = cosine + edge_attr[:, 9].cpu().numpy() + edge_attr[:, 1].cpu().numpy()
    is_self = edge_attr[:, 8].cpu().numpy() > 0.5
    frame = pd.DataFrame({"dst": dst, "strength": strength, "edge": np.arange(len(dst))})
    frame["rank"] = frame.groupby("dst", sort=False)["strength"].rank(method="first", ascending=False)
    keep = (frame["rank"].to_numpy() <= k) | is_self
    selected = frame["edge"].to_numpy()[keep]
    return edge_index[:, selected], edge_attr[selected]

def build_aml_edges(features, identities, timestamps):
    n = features.shape[0]
    norm = np.linalg.norm(features, axis=1, keepdims=True); norm[norm == 0] = 1.0
    features_norm = (features / norm).astype(np.float32)
    parts = []
    for column in IDENTITY_COLS:
        vals = identities[column]
        src, dst = _chain_pairs(vals, timestamps, EDGE_PER_TRANS, valid=_valid_id(vals))
        if len(src):
            parts.append((src, dst, REL_INDEX[column])); print(f"    {column}: {len(src):,} edges")
    use_sim = USE_SIMILARITY_EDGES and n <= SIM_MAX_NODES
    if use_sim:
        src, dst = knn_sim(features_norm, timestamps, identities["from_id"], K_SIM, device, SIM_BATCH)
        if len(src):
            parts.append((src, dst, REL_INDEX["similarity"])); print(f"    similarity(kNN k={K_SIM}): {len(src):,} edges")
    else:
        print(f"    similarity edges: DISABLED (N={n:,} > SIM_MAX_NODES={SIM_MAX_NODES:,} or flag off)")

    if parts:
        src_all = np.concatenate([p[0] for p in parts]); dst_all = np.concatenate([p[1] for p in parts])
        rel_all = np.concatenate([np.full(len(p[0]), p[2], np.int8) for p in parts])
        frame = pd.DataFrame({"src": src_all, "dst": dst_all})
        for rid in range(5): frame[f"r{rid}"] = (rel_all == rid).astype(np.float32)
        grouped = frame.groupby(["src", "dst"], sort=False).max().reset_index()
        src = grouped["src"].to_numpy(); dst = grouped["dst"].to_numpy()
        relation_flags = grouped[[f"r{i}" for i in range(5)]].to_numpy(np.float32)
        relation_count = relation_flags.sum(1, keepdims=True)
        delta_days = ((timestamps[dst] - timestamps[src]).astype(np.float32)) / 86400.0
        recency = np.stack([np.exp(-delta_days / float(t)) for t in DECAY_TAUS], 1).astype(np.float32)
        cosine = np.empty((len(src), 1), np.float32)
        chunk = 1_000_000
        for s0 in range(0, len(src), chunk):
            s1 = min(s0 + chunk, len(src))
            cosine[s0:s1, 0] = (features_norm[src[s0:s1]] * features_norm[dst[s0:s1]]).sum(1)
        edge_attr_np = np.concatenate(
            [recency, relation_flags, np.zeros((len(src), 1), np.float32), relation_count, cosine], 1
        ).astype(np.float32)
        edge_index = torch.tensor(np.stack([src, dst]), dtype=torch.long)
        edge_attr = torch.tensor(edge_attr_np)
    else:
        edge_index = torch.empty((2, 0), dtype=torch.long); edge_attr = torch.zeros((0, 11), torch.float32)

    loops = torch.arange(n, dtype=torch.long)
    self_attr = torch.zeros((n, 11), dtype=torch.float32)
    self_attr[:, 0:3] = 1.0; self_attr[:, 8] = 1.0; self_attr[:, 10] = 1.0
    edge_index = torch.cat([edge_index, torch.stack([loops, loops])], 1)
    edge_attr = torch.cat([edge_attr, self_attr], 0)
    print(f"    total edges incl self-loops: {edge_index.shape[1]:,}; edge_dim={edge_attr.shape[1]}")
    return edge_index, edge_attr

def _make_y_input(y, known_idx, n_classes=2):
    y_input = torch.full((len(y),), n_classes, dtype=torch.long)
    if len(known_idx):
        idx = torch.as_tensor(known_idx, dtype=torch.long); y_input[idx] = y.cpu()[idx]
    return y_input

class GTANEdge(nn.Module):
    def __init__(self, in_feats, hidden_dim=32, heads=4, n_layers=2, n_classes=2, dropout=0.2, edge_dim=11):
        super().__init__()
        self.n_classes = n_classes; width = hidden_dim * heads
        self.label_emb = nn.Embedding(n_classes + 1, in_feats, padding_idx=n_classes)
        self.feat_lin = nn.Linear(in_feats, width); self.label_lin = nn.Linear(in_feats, width)
        self.label_proc = nn.Sequential(nn.BatchNorm1d(width), nn.PReLU(), nn.Dropout(dropout),
                                        nn.Linear(width, in_feats))
        self.input_drop = nn.Dropout(dropout)
        self.convs = nn.ModuleList(); self.norms = nn.ModuleList(); dim = in_feats
        for _ in range(n_layers):
            self.convs.append(TransformerConv(dim, hidden_dim, heads=heads, concat=True, beta=True,
                                              dropout=dropout, edge_dim=edge_dim))
            self.norms.append(nn.LayerNorm(width)); dim = width
        self.act = nn.PReLU(); self.drop = nn.Dropout(dropout); self.emb_dim = dim
        self.head = nn.Sequential(nn.Linear(dim, dim), nn.BatchNorm1d(dim), nn.PReLU(),
                                  nn.Dropout(dropout), nn.Linear(dim, n_classes))
    def forward(self, x, edge_index, y_input, edge_attr=None):
        labels = self.input_drop(self.label_emb(y_input))
        hidden = x + self.label_proc(self.feat_lin(x) + self.label_lin(labels))
        for conv, norm in zip(self.convs, self.norms):
            hidden = self.drop(self.act(norm(conv(hidden, edge_index, edge_attr=edge_attr))))
        return self.head(hidden), hidden

def _loader(x, edge_index, edge_attr, sampling_weight, y_input, input_nodes, batch_size, neighbors, shuffle, y=None):
    data = Data(x=x.cpu().float(), edge_index=edge_index.cpu())
    data.edge_attr = edge_attr.cpu().float()
    if sampling_weight is not None: data.edge_weight = sampling_weight.cpu().float()
    data.yik = y_input.cpu().long()
    if y is not None: data.y = y.cpu().long()
    kwargs = {}
    if USE_WEIGHTED_SAMPLING: kwargs["weight_attr"] = "edge_weight"
    try:
        return NeighborLoader(data, num_neighbors=list(neighbors), input_nodes=torch.as_tensor(input_nodes),
                              batch_size=batch_size, shuffle=shuffle, **kwargs)
    except Exception as e:
        print(f"  [loader] {type(e).__name__}; uniform sampling fallback")
        return NeighborLoader(data, num_neighbors=list(neighbors), input_nodes=torch.as_tensor(input_nodes),
                              batch_size=batch_size, shuffle=shuffle)

def _mask_seed_labels(y_input_batch, seed_count):
    y = y_input_batch.clone(); y[:seed_count] = 2; return y

print("GTAN helpers ready")
```

```python
# ===== Approach B — reference selection, training, leakage-safe embeddings =====
def select_reference_and_query(train_idx, uid_values, seed):
    train_idx = np.asarray(train_idx, dtype=np.int64)
    if LABEL_PROTOCOL == "full" or REF_FRAC <= 0:
        return train_idx.copy(), train_idx.copy()   # reveal all train labels
    rng = np.random.default_rng(seed)
    uid = np.asarray(uid_values)
    train_uids = uid[train_idx]; valid = _valid_id(train_uids)
    cand = pd.unique(train_uids[valid])
    if len(cand):
        n_ref = min(len(cand), max(1, int(round(REF_FRAC * len(cand)))))
        ref_uids = rng.permutation(cand)[:n_ref]
        reference_idx = train_idx[np.isin(train_uids, ref_uids)]
    else:
        n_ref = min(len(train_idx), max(1, int(round(REF_FRAC * len(train_idx)))))
        reference_idx = np.sort(rng.permutation(train_idx)[:n_ref])
    query_idx = np.setdiff1d(train_idx, reference_idx, assume_unique=False)
    if len(query_idx) == 0: query_idx = train_idx.copy()
    return np.asarray(reference_idx, np.int64), np.asarray(query_idx, np.int64)

def train_gtan_edge(x, edge_index, edge_attr, y, query_idx, val_idx, reference_idx, in_feats, dev, verbose=True):
    y_cpu = y.cpu()
    query_idx = np.asarray(query_idx, np.int64); val_idx = np.asarray(val_idx, np.int64)
    reference_idx = np.asarray(reference_idx, np.int64)
    pos = float((y_cpu[torch.as_tensor(query_idx)] == 1).sum().clamp(min=1))
    neg = float((y_cpu[torch.as_tensor(query_idx)] == 0).sum().clamp(min=1))
    class_weight = torch.tensor([1.0, float(np.sqrt(neg / pos))], device=dev)
    sampling_weight = None
    if USE_WEIGHTED_SAMPLING:
        cosine = edge_attr[:, 10].clamp(0, 1)
        sampling_weight = (edge_attr[:, 9] + edge_attr[:, 1] + cosine + 1e-3).float()
    model = GTANEdge(in_feats, GTAN_HIDDEN, GTAN_HEADS, GTAN_LAYERS, 2, GTAN_DROP,
                     edge_dim=edge_attr.shape[1]).to(dev)
    opt = torch.optim.Adam(model.parameters(), lr=GTAN_LR, weight_decay=GTAN_WD)
    y_input = _make_y_input(y_cpu, reference_idx)
    tl = _loader(x, edge_index, edge_attr, sampling_weight, y_input, query_idx, GTAN_BATCH, GTAN_NEIGH, True, y=y)
    vl = _loader(x, edge_index, edge_attr, sampling_weight, y_input, val_idx, GTAN_BATCH, GTAN_NEIGH, False, y=y)
    best_auc, best_state, bad = -1.0, None, 0
    for ep in range(1, GTAN_EPOCHS + 1):
        model.train(); tot = nb = 0
        for b in tl:
            sc = b.batch_size; b = b.to(dev)
            opt.zero_grad()
            logits, _ = model(b.x, b.edge_index, _mask_seed_labels(b.yik, sc), b.edge_attr)
            loss = F.cross_entropy(logits[:sc], b.y[:sc], weight=class_weight)
            loss.backward(); nn.utils.clip_grad_norm_(model.parameters(), 2.0); opt.step()
            tot += float(loss); nb += 1
        model.eval(); vp = np.zeros(y.shape[0], np.float32)
        with torch.no_grad():
            for b in vl:
                sc = b.batch_size; b = b.to(dev)
                logits, _ = model(b.x, b.edge_index, b.yik, b.edge_attr)
                vp[b.n_id[:sc].cpu().numpy()] = F.softmax(logits[:sc], 1)[:, 1].cpu().numpy()
        va = roc_auc_score(y_cpu.numpy()[val_idx], vp[val_idx])
        if va > best_auc + GTAN_EARLY_STOP_MIN_DELTA:
            best_auc, best_state, bad = va, {k: v.detach().clone() for k, v in model.state_dict().items()}, 0
        else: bad += 1
        if verbose: print(f"   ep {ep:2d}/{GTAN_EPOCHS} loss={tot/max(nb,1):.4f} val_auc={va:.4f} best={best_auc:.4f}")
        if GTAN_EARLY_STOP_PATIENCE and bad >= GTAN_EARLY_STOP_PATIENCE:
            print(f"   early stop at epoch {ep}"); break
    if best_state: model.load_state_dict(best_state)
    print(f"   >>> GTAN best val AUC = {best_auc:.4f}")
    return model, sampling_weight

@torch.no_grad()
def _predict(model, x, edge_index, edge_attr, sampling_weight, y_input, input_nodes, dev):
    model.eval(); input_nodes = np.asarray(input_nodes, np.int64)
    prob = np.zeros(x.shape[0], np.float32); emb = np.zeros((x.shape[0], model.emb_dim), np.float32)
    loader = _loader(x, edge_index, edge_attr, sampling_weight, y_input, input_nodes, GTAN_BATCH, GTAN_NEIGH, False)
    for b in loader:
        sc = b.batch_size; b = b.to(dev)
        logits, hidden = model(b.x, b.edge_index, b.yik, b.edge_attr)
        nid = b.n_id[:sc].cpu().numpy()
        prob[nid] = F.softmax(logits[:sc], 1)[:, 1].cpu().numpy()
        emb[nid] = hidden[:sc].cpu().numpy()
    return prob, emb

def reference_safe_embeddings(model, x, edge_index, edge_attr, sampling_weight, y, reference_idx, dev):
    # Every non-reference node is scored with reference labels revealed (own label
    # masked); reference nodes are scored out-of-fold so their own label never leaks.
    n = x.shape[0]; prob = np.zeros(n, np.float32); emb = np.zeros((n, model.emb_dim), np.float32)
    reference_idx = np.asarray(reference_idx, np.int64)
    other = np.setdiff1d(np.arange(n), reference_idx, assume_unique=False)
    base_y = _make_y_input(y.cpu(), reference_idx)
    p, e = _predict(model, x, edge_index, edge_attr, sampling_weight, base_y, other, dev)
    prob[other] = p[other]; emb[other] = e[other]
    if len(reference_idx):
        rng = np.random.default_rng(REF_SEED); shuffled = rng.permutation(reference_idx)
        for held in np.array_split(shuffled, REFERENCE_OOF_FOLDS):
            if len(held) == 0: continue
            revealed = np.setdiff1d(reference_idx, held, assume_unique=False)
            hy = _make_y_input(y.cpu(), revealed)
            p, e = _predict(model, x, edge_index, edge_attr, sampling_weight, hy, held, dev)
            prob[held] = p[held]; emb[held] = e[held]
    return prob, emb

print("GTAN train / embedding routines ready")
```

### 5a · Approach B — build graph, train GTAN (graph alone)

One graph is built over **all** sub-sampled transactions (edges use only
timestamps + shared accounts, never labels). Training supervises the query-train
nodes, early-stops on val, and reports test. Edges always run earlier→later in
time, so a node only ever aggregates from its past → temporally valid.

```python
# ===== Approach B — build graph + train =====
node_idx = np.arange(len(tx))
train_local = node_idx[tr_mask]; val_local = node_idx[va_mask]; test_local = node_idx[te_mask]
identities = {c: pd.factorize(tx[c].to_numpy())[0] for c in IDENTITY_COLS}

print("building transaction graph ...")
B_edge_index, B_edge_attr = build_aml_edges(X_std, identities, time_arr)
if PRUNE_TOPK:
    before = B_edge_index.shape[1]
    B_edge_index, B_edge_attr = prune_topk_per_destination(B_edge_index, B_edge_attr, PRUNE_K)
    print(f"    top-{PRUNE_K} prune: {before:,} -> {B_edge_index.shape[1]:,} edges")

reference_local, query_local = select_reference_and_query(train_local, identities["from_id"], REF_SEED)
print(f"label protocol = {LABEL_PROTOCOL} | reference={len(reference_local):,} query={len(query_local):,} "
      f"val={len(val_local):,} test={len(test_local):,}")

xB = torch.tensor(X_std)
yB = torch.tensor(y, dtype=torch.long)
gtan_model, B_sw = train_gtan_edge(xB, B_edge_index, B_edge_attr, yB,
                                   query_local, val_local, reference_local, NF, device, verbose=True)

# graph-alone test predictions: test nodes are scored with reference labels revealed
test_y_input = _make_y_input(yB.cpu(), reference_local)
B_prob, _ = _predict(gtan_model, xB, B_edge_index, B_edge_attr, B_sw, test_y_input, test_local, device)
report("3a · GTAN (transaction-graph) alone", y[te_mask], B_prob[test_local])
```

### 5b · Approach B — Graph + XGBoost

Extract **leakage-safe** GTAN node embeddings for every transaction
(`reference_safe_embeddings`), standardize them on the train rows, concatenate
with the raw tabular features, and feed XGBoost.

```python
# ===== Approach B — GTAN embeddings + XGBoost =====
print("extracting reference-safe GTAN embeddings ...")
_, B_emb = reference_safe_embeddings(gtan_model, xB, B_edge_index, B_edge_attr, B_sw, yB, reference_local, device)
emu = B_emb[tr_mask].mean(0); esd = B_emb[tr_mask].std(0) + 1e-6
B_emb_std = ((B_emb - emu) / esd).astype(np.float32)
B_feats = np.concatenate([B_emb_std, X_raw], axis=1).astype(np.float32)
print("Approach-B XGB feature dim:", B_feats.shape[1], "(= gtan_emb + raw)")
prob, _ = run_xgb(B_feats[tr_mask], y[tr_mask], B_feats[va_mask], y[va_mask],
                  B_feats[te_mask], y[te_mask], "B: GTAN + XGB")
report("3b · GTAN + XGB", y[te_mask], prob)
```
